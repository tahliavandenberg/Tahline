# Bepaling van soortelijke warmte van water


# Introductie

Water heeft een enorme buffercapaciteit: je moet veel energie toevoegen om de temperatuur van water een graad te verwarmen. In dit practicum gaan we de soortelijke warmte van water bepalen door een bekende hoeveelheid water te verwarmen met een bekende hoeveelheid energie, en de temperatuurstijging te meten.

# Theorie
De soortelijke warmte van water  4186 J/kg * K. Soortelijke warmte van een stof is een eigenschap. Zo heeft elke stof een eigen soortelijke warmte. Deze eigenschap geeft aan hoeveel warmte er nodig is om een bepaalde stof met een aantal hoeveelheid graden te verhogen. Met behulp van deze formules kan de soortelijke warmte bepaald worden. 
$$
    Q = c \cdot m \cdot \Delta T 
$$
$$
    P = U \cdot I
$$
$$
    Q = P \cdot t
$$
$$
    \rho = \frac {m}{V}
$$

# Resultaten


````python
# Hier de data en de analyse
import numpy as np
import matplotlib.pyplot as plt


temp = np.array([ 21.4 , 21.5 , 21.6, 21.7, 21.9, 22.0, 22.2, 22.5, 22.6, 22.8, 23, 24.3, 24.5]) + 273.15 #Kelvin
tijd = np.array([ 35, 40, 50, 57, 60, 66, 74, 83, 90, 96, 103, 134, 145]) #seconde
volume = 350 # mL
U = 20 #V
I = 2 #A
````

````python
# Sla figuren op met  
# 
plt.figure()
plt.plot(tijd, temp, '.', label = 'data')
plt.xlabel('$t$ (s)')
plt.ylabel('$T$(K)')
plt.show()
# plt.savefig("figures/naam.png", dpi=450)


     
````

````python
rho_water = 0.997 #g/mL
massa = (volume * rho_water) / 1000 #kg
P = U * I #Watt
t = tijd[-1] - tijd[0]
Q = P * t
delta_T = temp[-1] - temp[0]
# Q = c * m * delta_T
c = Q / (massa * delta_T)
print('De soortelijke warmte van water is', np.round(c, 1), 'J/(kg * K)')
````
